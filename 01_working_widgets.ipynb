### select the environment

```python
dbutils.widgets.remove("env")
```

```python
dbutils.widgets.text("env", "xyz", "environment")
env = dbutils.widgets.get("env")
print(f"Selected environment : {env}")
```

### dropdown input

```python
dbutils.widgets.dropdown("environment", "dev", 
    ['dev','test','prod'],
    "select environment")
```

```python
env = dbutils.widgets.get("environment")
print(f"Selected environment : {env}")
```

### Multiselect Widget


```python
dbutils.widgets.multiselect("contries", "India", 
    ['India','USA','UK'],
    "select country")
```

```python
countries = dbutils.widgets.get("contries")
print(f"Selected countries : {countries}")
```

```python
dbutils.widgets.dropdown("run_job", "False", 
    ['True','False'],
    "run_job")
run_job = dbutils.widgets.get("run_job")
print(f"run_job : {run_job}")
```
