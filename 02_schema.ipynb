```python
from pyspark.sql.types import *

schema = StructType([
    StructField("EmpId", IntegerType(), True),
    StructField("Name", StringType(), True),
    StructField("Salary", DoubleType(), True)
])

df = spark.read.csv("/FileStore/employees.csv",
                    header=True,
                    schema=schema)
```
